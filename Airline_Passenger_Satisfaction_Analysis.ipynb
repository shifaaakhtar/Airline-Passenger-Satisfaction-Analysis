```python
#delay threshold

df["delay_bucket"] = pd.cut(
    df["arrival_delay_in_minutes"],
    bins=[0, 15, 60, 120, 1000],
    labels=[
        "0-15 mins",
        "15-60 mins",
        "60-120 mins",
        "120+ mins"
    ]
)

pd.crosstab(
    df["delay_bucket"],
    df["satisfaction"],
    normalize="index"
)
```

```python
df["Delay Duration Threshold"] = pd.cut(
    df["arrival_delay_in_minutes"],
    bins=[-1, 0, 15, 60, 120, 10000],
    labels=["No Delay", "0–15", "15–60", "60–120", "120+"]
)

delay_satisfaction = pd.crosstab(
    df["Delay Duration Threshold"],
    df["satisfaction"],
    normalize="index"
) * 100

delay_satisfaction.plot(
    kind="bar",
    stacked=True
)

plt.title("Satisfaction Rate by Delay Duration")
plt.ylabel("Percentage (%)")
plt.show()
```
